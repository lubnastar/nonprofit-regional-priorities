# أين يجب أن تتجه موارد العمل الخيري بين مناطق المملكة؟

**سؤال القرار:** أين ينبغي للمانحين والجمعيات إعطاء الأولوية عند توزيع الموارد بين مناطق المملكة؟

**السؤال التحليلي:** ما المناطق التي تجمع بين ارتفاع مؤشرات الهشاشة الاقتصادية وانخفاض كثافة المنظمات غير الربحية؟

### المفاهيم الأساسية

**الهشاشة الاقتصادية:** تشير إلى ضعف قدرة الأسر في المنطقة على تلبية احتياجاتها. ويُقاس هذا المفهوم من خلال ثلاثة مؤشرات:
- **نسبة مستفيدي الضمان الاجتماعي من السكان السعوديين:** ارتفاع النسبة يشير إلى زيادة الاعتماد على الدعم الحكومي.
- **وسيط الدخل الشهري المتاح للفرد السعودي:** انخفاض الوسيط يشير إلى ارتفاع الهشاشة الاقتصادية.
- **حصة الغذاء من إنفاق الأسرة:** تميل الأسر الأقل دخلًا إلى تخصيص نسبة أكبر من إنفاقها للاحتياجات الغذائية.

**كثافة المنظمات غير الربحية:** عدد الجمعيات والمؤسسات الأهلية لكل 100 ألف نسمة من إجمالي السكان. ويُستخدم هذا المؤشر لتقدير مدى توافر البنية المؤسسية للعمل الخيري في المنطقة. ومع ذلك، فإن عدد المنظمات لا يعكس بالضرورة حجم خدماتها أو نطاقها الجغرافي؛ فقد تقدم المنظمة خدماتها خارج المنطقة التي يقع فيها مقرها (القسم 10).

### الخلاصة التنفيذية

- **الإجابة عن السؤال التحليلي:** تتصدر منطقة نجران قائمة الأولويات بفارق واضح، تليها منطقة جازان. وتجمع المنطقتان بين ارتفاع الهشاشة الاقتصادية وانخفاض كثافة المنظمات غير الربحية.
- **مجموعة الأولوية الثانية:** المدينة المنورة وتبوك ومكة المكرمة. تعاني هذه المناطق من انخفاض كثافة المنظمات مقارنة بحجم سكانها، رغم أن مستوى الهشاشة الاقتصادية فيها يساوي المتوسط أو يقل عنه. ولذلك، فهي مناسبة للمانحين الذين يستهدفون سد الفجوات في تقديم الخدمات.
- **منهجية تحديد الأولويات:** رُتبت المناطق وفق الفجوة بين مؤشر الهشاشة الاقتصادية ومؤشر كثافة المنظمات. ثم اختيرت المناطق التي بقيت ضمن المراتب الخمس الأولى في كل اختبارات الحساسية الست إلا واحدًا على الأكثر، وقُسمت حسب مصدر الفجوة: الهشاشة الاقتصادية أم قلة المنظمات.
- **متانة النتائج:** حافظت المناطق الخمس نفسها على المراتب المتقدمة عبر السيناريوهات الأربعة وتجارب الحساسية الست (القسم 8). لكن هذه الاختبارات تقيس استقرار النتائج عند تغيير طريقة الحساب، ولا تثبت دقة البيانات الأصلية (القسم 10).

## 2. تحميل البيانات

جداول مسح الدخل والإنفاق منشورة في ملف Excel بأربعة جداول متفرقة، فنستخرج منها جدولًا واحدًا، صف لكل منطقة، من الملف الخام مباشرة: (11-2) وسيط الدخل، و(10-2) متوسط الدخل، و(9-3) متوسط الإنفاق، و(16-4) حصص الإنفاق. نأخذ الاسم العربي من أول سطر في خلية المنطقة، ونحذف صف الإجمالي.

```python
import openpyxl

wb = openpyxl.load_workbook(RAW / "hies_2023_publication_tables.xlsx", data_only=True, read_only=True)


def region_table(sheet, prefix):
    """جدول أعمدته سعوديون وغير سعوديين وإجمالي: صف لكل منطقة بلا صف الإجمالي."""
    rows = []
    for name, *vals in wb[sheet].iter_rows(values_only=True):
        if isinstance(name, str) and isinstance(vals[0], (int, float)):
            region = name.split("\n")[0].strip()
            if region != "الإجمالي":
                rows.append((region, *vals[:3]))
    cols = [f"{prefix}_{n}" for n in ("saudi", "non_saudi", "total")]
    return pd.DataFrame(rows, columns=["region_ar", *cols]).set_index("region_ar")


# (16-4): المناطق أعمدة ومجموعات الإنفاق صفوف، فنقلبه ليصير صفًا لكل منطقة
SHARE_NAMES = [
    "food", "tobacco", "clothing", "housing_utilities", "furnishings", "health", "transport",
    "information_communication", "recreation", "education", "restaurants_accommodation",
    "insurance_finance", "personal_care_other",
]
rows_164 = list(wb["16-4"].iter_rows(values_only=True))
header = next(r for r in rows_164 if isinstance(r[1], str) and "Riyadh" in r[1])
share_rows = [r for r in rows_164 if isinstance(r[1], (int, float))][: len(SHARE_NAMES)]
shares = pd.DataFrame(
    [r[1:14] for r in share_rows],
    index=[f"share_{n}" for n in SHARE_NAMES],
    columns=[h.split("\n")[0].strip() for h in header[1:14]],
).T.rename_axis("region_ar")

hies_raw = (
    region_table("11-2", "income_median")
    .join(region_table("10-2", "income_mean"))
    .join(region_table("9-3", "spend_mean"))
    .join(shares)
    .reset_index()
)
assert len(hies_raw) == 13 and hies_raw.notna().all().all()
```

```python
# السطر الأول في ملف الجمعيات عنوان وليس رأس أعمدة، فنتجاوزه
charities_raw = pd.read_csv(RAW / "nonprofit_organizations.csv", skiprows=1, encoding="utf-8-sig")
population_raw = pd.read_csv(RAW / "sama_population_by_region_2010_2022.csv")
saudi_raw = pd.read_csv(RAW / "gastat_saudi_population_by_region_2010_2022.csv", encoding="utf-8-sig")
social_raw = pd.read_csv(RAW / "sama_social_security_by_region_2005_2022.csv")

print("charities:", charities_raw.shape)
print("population:", population_raw.shape)
print("saudi_population:", saudi_raw.shape)
print("social_security:", social_raw.shape)
print("hies:", hies_raw.shape)
```

### 3.1 توحيد أسماء المناطق

تختلف أسماء المناطق بين ملفات البيانات؛ لذلك يُستخدم جدول ربط موحّد يعتمد المتغير `region_id`، الذي يتراوح بين 1 و13، ويتطابق مع العمود `Province ID` في ملفات البنك المركزي السعودي.

| العمود | الوصف |
|---|---|
| `region_ar` | الاسم العربي للمنطقة كما ورد في مسح الدخل والإنفاق وملف السكان السعوديين |
| `region_en` | الاسم الإنجليزي الموحّد المستخدم في الرسوم |
| `charities_name` | اسم المنطقة كما ورد في ملف الجمعيات والمؤسسات |
| `sama_name` | اسم المنطقة كما ورد في ملفات البنك المركزي السعودي |

```python
region_map = pd.DataFrame(
    [
        (1, "الرياض", "Riyadh", "منطقة الرياض", "Al-Riyadh"),
        (2, "مكة المكرمة", "Makkah", "منطقة مكة المكرمة", "Makkah Al-Mokarramah"),
        (3, "المدينة المنورة", "Madinah", "منطقة المدينة المنورة", "Al-Madinah Al-Monawarah"),
        (4, "القصيم", "Qassim", "منطقة القصيم", "Al-Qaseem"),
        (5, "المنطقة الشرقية", "Eastern Province", "المنطقة الشرقية", "Eastern Region"),
        (6, "عسير", "Asir", "منطقة عسير", "Aseer"),
        (7, "تبوك", "Tabuk", "منطقة تبوك", "Tabouk"),
        (8, "حائل", "Hail", "منطقة حائل", "Hail"),
        (9, "الحدود الشمالية", "Northern Borders", "منطقة الحدود الشمالية", "Northern Borders"),
        (10, "جازان", "Jazan", "منطقة جازان", "Jazan"),
        (11, "نجران", "Najran", "منطقة نجران", "Najran"),
        (12, "الباحة", "Al-Baha", "منطقة الباحة", "Al-Baha"),
        (13, "الجوف", "Al-Jouf", "منطقة الجوف", "Al-Jouf"),
    ],
    columns=["region_id", "region_ar", "region_en", "charities_name", "sama_name"],
)
region_map
```

**التحقق من صحة الربط:** يُجرى فحصان، ويوقف التنفيذ إذا ظهر أي تعارض:
1. **معرّفات المناطق:** التحقق من تطابق أرقام المناطق في ملفات البنك المركزي مع معرّفات جدول الربط؛ لتجنب إسناد البيانات إلى منطقة غير صحيحة.
2. **أسماء المناطق:** التأكد من وجود جميع الأسماء الواردة في الملفات ضمن جدول الربط. وعند وجود اختلاف، تعرض رسالة الخطأ الأسماء غير المتطابقة.

```python
# 1. أرقام المناطق في SAMA تطابق region_id
for name, df in [("sama_population", population_raw), ("sama_social_security", social_raw)]:
    ids = df[["Province ID", "Province"]].drop_duplicates().merge(region_map, left_on="Province", right_on="sama_name")
    if len(ids) != 13 or not (ids["Province ID"] == ids["region_id"]).all():
        raise ValueError(f"{name}: أرقام المناطق لا تطابق region_id")

# 2. كل اسم منطقة في كل ملف موجود في الجدول
names_in_files = {
    "nonprofit_organizations": (charities_raw["المنطقة"], "charities_name"),
    "sama_population": (population_raw["Province"], "sama_name"),
    "gastat_saudi_population": (saudi_raw["المنطقة"], "region_ar"),
    "sama_social_security": (social_raw.loc[social_raw["Province"] != "Grand Total", "Province"], "sama_name"),
    "hies_2023_publication_tables": (hies_raw["region_ar"], "region_ar"),
}
for name, (found, col) in names_in_files.items():
    found, expected = set(found), set(region_map[col])
    if found != expected:
        raise ValueError(f"{name}: زائد {found - expected} | ناقص {expected - found}")

print("الربط سليم: أرقام المناطق في SAMA مطابقة، وكل الأسماء في الملفات الخمسة موجودة في الجدول.")
```

اجتاز الربط كلا الفحصين، ما يؤكد اتساق أسماء المناطق ومعرّفاتها بين الملفات.

### 3.2 تنظيف بيانات الجمعيات والمؤسسات

تُحذف الصفوف المتطابقة بالكامل فقط، مع الاحتفاظ بالأسماء المبتورة (المنتهية بعلامة `…`) ووضع علامة عليها، حتى لا تُستبعد جهات قد تكون بياناتها غير مكتملة.

```python
charities = (
    charities_raw.drop_duplicates()
    .rename(columns={"رقم التسجيل": "reg_no", "اسم الجمعية": "name", "نوع الجهة": "org_type", "المنطقة": "charities_name"})
    .merge(region_map[["region_id", "charities_name"]], on="charities_name", how="left", validate="many_to_one")
)
print("صفوف متطابقة محذوفة:", len(charities_raw) - len(charities))

charities["org_type"] = charities["org_type"].map({"جمعية أهلية": "association", "مؤسسة أهلية": "foundation"})
charities["name_truncated"] = charities["name"].str.endswith("…")
assert charities[["region_id", "org_type"]].notna().all().all()

charities = charities[["region_id", "org_type", "reg_no", "name", "name_truncated"]]
charities.to_csv(CLEAN / "charities.csv", index=False, encoding="utf-8-sig")

charity_counts = pd.crosstab(charities["region_id"], charities["org_type"]).reset_index().rename_axis(columns=None)
charity_counts["total"] = charity_counts["association"] + charity_counts["foundation"]
charity_counts
```

لم تُحذف أي صفوف في عملية التنظيف، وربطت جميع الجهات بالمناطق الصحيحة. ويُعرض توزيع الجمعيات والمؤسسات حسب المنطقة في القسم 4.1.

### 3.3 تنظيف بيانات السكان

```python
pop_wide = population_raw.pivot(index="Province ID", columns="Year", values="Population")
saudi_wide = (
    saudi_raw.merge(region_map[["region_id", "region_ar"]], left_on="المنطقة", right_on="region_ar", validate="one_to_one")
    .set_index("region_id")
)

population = pd.DataFrame({
    "population": pop_wide[REF_YEAR].astype("int64"),
    "saudi_population": saudi_wide[str(REF_YEAR)].astype("int64"),
    "saudi_population_avg": saudi_wide[[str(y) for y in AVG_YEARS]].mean(axis=1),
}).rename_axis("region_id").reset_index()
assert (population["saudi_population"] < population["population"]).all()  # السعوديون جزء من كل السكان
population
```

ينتج عن ذلك ثلاثة متغيرات لكل منطقة:
- `population`: إجمالي السكان في 2022، ويُستخدم مقامًا لحساب كثافة المنظمات.
- `saudi_population`: عدد السكان السعوديين في 2022، ويُستخدم مقامًا لحساب نسبة مستفيدي الضمان في السنة نفسها.
- `saudi_population_avg`: متوسط عدد السكان السعوديين خلال 2020-2022، ويُستخدم مقامًا لحساب نسبة الضمان المبنية على المتوسط، بما يضمن توافق الفترة الزمنية للبسط والمقام.

### 3.4 الضمان الاجتماعي

قبل حذف صف `Grand Total` نتحقق أنه يساوي مجموع المناطق في كل سنة.

```python
social = social_raw.rename(columns={"Province ID": "region_id", "Year": "year", "Beneficiaries": "beneficiaries"})
is_total = social["Province"] == "Grand Total"

# صف الإجمالي يجب أن يساوي مجموع المناطق قبل حذفه
assert (social[~is_total].groupby("year")["beneficiaries"].sum() == social[is_total].set_index("year")["beneficiaries"]).all()

social = (
    social.loc[~is_total, ["region_id", "year", "beneficiaries"]]
    .astype({"beneficiaries": "int64"})
    .sort_values(["region_id", "year"], ignore_index=True)
)
social.to_csv(CLEAN / "social_security.csv", index=False)
print(social.shape, social["year"].min(), "-", social["year"].max())
```

## 4. التحليل الاستكشافي

**السؤال:** ماذا تقول البيانات قبل بناء المؤشرات، وما القرارات التي تحسمها؟

نتعرف على البيانات النظيفة قبل بناء المؤشرات. كل قسم يجيب عن سؤال، وثلاثة منها تحسم قرارات في طريقة الحساب.

```python
# إعدادات الرسوم: لون أزرق للفئة الأولى وبرتقالي للثانية، وخطوط شبكة خفيفة
BLUE, ORANGE, INK, MUTED, GRID = "#2a78d6", "#eb6834", "#0b0b0b", "#898781", "#e1e0d9"
OTHER = "#b4b2a9"  # رسوم الأقسام 6-9: البرتقالي للهشاشة، والأزرق لنقص المنظمات، والرمادي للباقي
plt.rcParams.update({
    "figure.dpi": 110, "savefig.dpi": 150, "savefig.bbox": "tight", "font.size": 10,
    "axes.edgecolor": "#c3c2b7", "axes.labelcolor": INK, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6, "axes.axisbelow": True,
    "xtick.color": MUTED, "ytick.color": MUTED, "legend.frameon": False,
})

# الرسوم بالأسماء الإنجليزية لأن matplotlib لا يدعم الحروف العربية المتصلة
names = region_map.set_index("region_id")["region_en"]


def bar_labels(ax, values, fmt="{}", pad=0, fontsize=8.5):
    """قيمة كل عمود أفقي مكتوبة بجانبه."""
    for yi, v in enumerate(values):
        ax.text(v + pad, yi, fmt.format(v), va="center", fontsize=fontsize, color=INK)


def group_colors(regions, fragility_led, density_led):
    """برتقالي إذا كان مصدر الفجوة الهشاشة، وأزرق إذا كان نقص المنظمات، ورمادي للباقي."""
    return [ORANGE if r in fragility_led else BLUE if r in density_led else OTHER for r in regions]


def clean_y(*axes):
    """أعمدة أفقية: بلا خطوط شبكة أفقية ولا شرطات على محور المناطق."""
    for ax in axes:
        ax.grid(axis="y", visible=False)
        ax.tick_params(axis="y", length=0)


def show_table(df, highlight=(), region_col="region_en", gradient=(), reverse=(), precision=2):
    """جدول منسق: اسم منطقة الأولوية بخط عريض وخلفية برتقالية، وتدرج لوني على أعمدة مختارة.
    gradient: كلما ارتفعت القيمة غمق اللون. reverse: العكس (مثل الرتب، 1 = الأغمق)."""
    sty = df.style.hide(axis="index").format(precision=precision, thousands=",")
    for cols, cmap in [(gradient, "Oranges"), (reverse, "Oranges_r")]:
        cols = [c for c in cols if c in df.columns]
        if cols:
            sty = sty.background_gradient(subset=cols, cmap=cmap, axis=None)
    if highlight:
        sty = sty.map(lambda r: "font-weight:bold; background-color:#fde3d7" if r in highlight else "", subset=[region_col])
    display(sty)


def headline(fig, title, subtitle, y, sub_y):
    """عنوان يقول الخلاصة، وتحته سطر يشرح ما في الرسم."""
    fig.suptitle(title, x=0.125, y=y, ha="left", fontsize=12.5, color=INK, fontweight="bold")
    fig.text(0.125, sub_y, subtitle, ha="left", fontsize=9.5, color=MUTED)
```

### 4.1 توزيع الجمعيات

- هل تتركز المؤسسات في مناطق معينة؟ (يساعد في قرار: **الجمعيات والمؤسسات معًا أم الجمعيات فقط**)

```python
ch = charity_counts.assign(region=charity_counts["region_id"].map(names))
ch["foundation_pct"] = (ch["foundation"] / ch["total"] * 100).round(1)
ch["share_of_all_pct"] = (ch["total"] / ch["total"].sum() * 100).round(1)
ch = ch.sort_values("total", ascending=False)
print(ch[["region", "association", "foundation", "total", "foundation_pct", "share_of_all_pct"]].to_string(index=False))

riyadh = ch.loc[ch["region_id"] == 1].iloc[0]
riyadh_fdn_share = riyadh["foundation"] / ch["foundation"].sum()
print(f"\nالرياض: {riyadh['share_of_all_pct']}% من كل الجهات، و{riyadh_fdn_share:.0%} من المؤسسات")
print(f"مناطق بلا مؤسسات: {', '.join(ch.loc[ch['foundation'] == 0, 'region'])}")

# لوحتان بنفس ترتيب المناطق: الجمعيات، ثم المؤسسات بمقياس خاص بها حتى يظهر تركّزها
d = ch.sort_values("total")
y = range(len(d))
fig, (ax_a, ax_f) = plt.subplots(1, 2, figsize=(9.5, 5.2), sharey=True, gridspec_kw={"width_ratios": [2.6, 1], "wspace": 0.08})
for ax, col, color, scale in [(ax_a, "association", BLUE, 1.12), (ax_f, "foundation", ORANGE, 1.3)]:
    ax.barh(y, d[col], color=color, height=0.62)
    ax.set_title(col.capitalize() + "s", loc="left", fontsize=10, color=color, fontweight="bold")
    ax.set_xlim(0, d[col].max() * scale)
    ax.set_xlabel("Number of organizations", color=MUTED, fontsize=9)
bar_labels(ax_a, d["association"], pad=8)
for yi, f in enumerate(d["foundation"]):  # "none" بدل عمود فارغ لا يُلاحَظ
    if f == 0:
        ax_f.text(1.5, yi, "none", va="center", fontsize=8, color=MUTED, style="italic")
    else:
        ax_f.text(f + 3, yi, f"{f}", va="center", fontsize=8.5, color=INK)

# عمود المجموع وحصة المنطقة من كل الجهات، وهو أساس الترتيب
ax_f.annotate("Total", (1.06, 1.0), xycoords="axes fraction", xytext=(0, 6), textcoords="offset points",
              va="baseline", fontsize=10, color=INK, fontweight="bold")  # نفس ارتفاع عنواني اللوحتين
for yi, (t, share) in enumerate(zip(d["total"], d["share_of_all_pct"])):
    ax_f.text(1.06, yi, f"{t}", transform=ax_f.get_yaxis_transform(), va="center", fontsize=8.5, color=INK, fontweight="bold")
    ax_f.text(1.25, yi, f"{share:.0f}%", transform=ax_f.get_yaxis_transform(), va="center", fontsize=8.5, color=MUTED)

clean_y(ax_a, ax_f)
ax_f.spines["left"].set_visible(False)
ax_a.set_yticks(list(y), d["region"], color=INK)
headline(fig, f"Riyadh hosts {riyadh['share_of_all_pct']:.0f}% of all non-profits and {riyadh_fdn_share:.0%} of foundations",
         "Registered non-profit organizations by region of head office, sorted by total", y=1.0, sub_y=0.935)
fig.savefig(FIGS / "eda_01_charities_by_region.png")
plt.show()
```

**اللوحة اليسرى** (الجمعيات): الرياض ومكة في المقدمة بفارق كبير، وباقي المناطق متقاربة نسبيًا.

```python
national = social.groupby("year")["beneficiaries"].sum()
print(national.loc[2018:].to_string())

# هل ترتيب المناطق ثابت بين السنوات الأخيرة؟ نقارن النسب لا الأعداد، لأن الأعداد تتبع حجم المنطقة
wide = social.pivot(index="region_id", columns="year", values="beneficiaries")
rates = wide[AVG_YEARS] / saudi_wide[[str(y) for y in AVG_YEARS]].set_axis(AVG_YEARS, axis=1) * 100
for y in (2020, 2021):
    print(f"ارتباط ترتيب نسبة الضمان {y} مقابل {REF_YEAR}: {rates[y].corr(rates[REF_YEAR], method='spearman'):.2f}")

recent = wide[[2020, 2021, 2022]].assign(region=names, change_21_22_pct=(wide[2022] / wide[2021] - 1) * 100)
print(recent.round(1).sort_values("change_21_22_pct").to_string())

# ---- الرسم: المملكة عبر السنوات (يسار)، وكل منطقة 2020-2022 بمؤشر 2020 = 100 (يمين)
peak_year = national.idxmax()
drop_pct = (national[REF_YEAR] / national[peak_year] - 1) * 100
fig, (ax_t, ax_r) = plt.subplots(1, 2, figsize=(11, 4.8), gridspec_kw={"width_ratios": [2.1, 1], "wspace": 0.28})

# فترة الانتقال إلى النظام المطور: من بدء التقديم (نوفمبر 2021) إلى إغلاق المنصة القديمة (مايو 2022)
ax_t.axvspan(2021 + 10 / 12, 2022 + 4 / 12, color=ORANGE, alpha=0.12, linewidth=0)
ax_t.text(2021 + 10 / 12, 0.15, "transition to\nnew system", fontsize=8, color=ORANGE, ha="right", va="bottom")
ax_t.plot(national.index, national / 1e6, color=BLUE, linewidth=2, marker="o", markersize=4, zorder=3)
for yr, dy in [(peak_year, 8), (REF_YEAR, -16)]:
    ax_t.annotate(f"{national[yr] / 1e6:.2f}M", (yr, national[yr] / 1e6), textcoords="offset points",
                  xytext=(0, dy), ha="center", fontsize=8.5, color=INK, fontweight="bold")
ax_t.set_title("All regions, 2005–2022 (millions)", loc="left", fontsize=10, color=INK)
ax_t.set_xticks(range(2005, 2023, 2))
ax_t.set_ylim(0, national.max() / 1e6 * 1.15)

index = wide[[2020, 2021, 2022]].div(wide[2020], axis=0) * 100
nat_index = national.loc[[2020, 2021, 2022]] / national[2020] * 100
for rid, row in index.iterrows():
    is_riyadh = rid == 1
    ax_r.plot(row.index, row.values, color=ORANGE if is_riyadh else BLUE, alpha=1 if is_riyadh else 0.25,
              linewidth=2.2 if is_riyadh else 1.2, marker="o" if is_riyadh else None, markersize=4, zorder=3 if is_riyadh else 2)
ax_r.plot(nat_index.index, nat_index.values, color=INK, linewidth=2, linestyle="--", zorder=4)
for label, value, color, weight in [("Riyadh", index.loc[1, 2022], ORANGE, "bold"), ("All regions", nat_index[2022], INK, "normal")]:
    ax_r.annotate(label, (2022, value), textcoords="offset points", xytext=(6, 0), va="center", fontsize=8.5, color=color, fontweight=weight)
ax_r.text(2020.05, index[2021].max(), "other regions", fontsize=8, color=BLUE, alpha=0.8, va="top")
ax_r.axhline(100, color=MUTED, linewidth=0.8)
ax_r.set_xticks([2020, 2021, 2022])
ax_r.set_xlim(2019.85, 2022.55)
ax_r.set_title("Each region, 2020 = 100", loc="left", fontsize=10, color=INK)

headline(fig, f"Beneficiaries fell {abs(drop_pct):.0f}% from {peak_year} to {REF_YEAR}, the transition year to the new system",
         "Social security beneficiaries. Every region fell in 2022, so regional ranks barely moved", y=1.03, sub_y=0.955)
fig.savefig(FIGS / "eda_03_social_security_trend.png")
plt.show()
```

### 4.4 الدخل والإنفاق
- وسيط دخل الفرد السعودي في كل منطقة.
- حصة الغذاء من الإنفاق حسب المنطقة.
- هل المناطق الأقل دخلًا تنفق حصة أكبر على الغذاء؟

## 5. بناء المؤشرات

جدول واحد، صف لكل منطقة.

- الدخل للسعوديين، ونسبة الضمان من السعوديين، لأن مستفيدي الضمان سعوديون.
- الكثافة من كل السكان، لأن الجمعيات تخدم كل من في المنطقة.
- هامش الدخل خارج المؤشر المركب، ويُجرَّب في اختبار الحساسية فقط.
- **سنة الضمان ونوع الجهات: نجرّب الخيارين في كل منهما**، فيكون عندنا 4 سيناريوهات في القسم 7.

**مؤشرات الهشاشة** (3 مؤشرات):

|---|---|---|
| `ss_rate_2022` | مستفيدو الضمان 2022 ÷ السعوديون 2022 × 100 | أعلى = أهش |
| `ss_rate_avg` | متوسط المستفيدين 2020-2022 ÷ متوسط السعوديين 2020-2022 × 100 | أعلى = أهش |

**وحدة المستفيدين:** «عدد المستفيدين» يُحتسب أفرادًا (العائل والتابعون)، فنعامله كأفراد ونقسمه على عدد الأفراد السعوديين.

**كثافة المنظمات** (لكل 100 ألف نسمة من كل السكان، 2022):

| المؤشر | الجهات |
|---|---|
| `density_all` | الجمعيات والمؤسسات |
| `density_assoc` | الجمعيات فقط |

```python
# نبدأ من الجدول الموحد (3.6)
ind = regions.drop(columns="region_ar").copy()

# الهشاشة: المستفيدون سعوديون، فالمقام السعوديون
ind["ss_rate_2022"] = ind["beneficiaries_2022"] / ind["saudi_population"] * 100
ind["ss_rate_avg"] = ind["beneficiaries_avg"] / ind["saudi_population_avg"] * 100  # متوسط على متوسط للسنوات نفسها

# الكثافة: الجمعيات تخدم كل السكان، فالمقام كل السكان
ind["density_all"] = ind["total"] / ind["population"] * 1e5
ind["density_assoc"] = ind["association"] / ind["population"] * 1e5

# لاختبار الحساسية فقط
ind["margin_saudi"] = ind["income_mean_saudi"] - ind["spend_mean_saudi"]

assert len(ind) == 13 and ind.notna().all().all()

cols = ["region_en", "ss_rate_2022", "ss_rate_avg", "income_median_saudi", "food_share", "density_all", "density_assoc"]
print(ind[cols].round(2).sort_values("density_all").to_string(index=False))
```

## 6. مؤشر الهشاشة المركب

**السؤال:** كيف نجمع المؤشرات الثلاثة في رقم واحد، وما المنطقة الأهش؟

- نحوّل كل مؤشر إلى درجة معيارية (z-score): كم انحرافًا معياريًا تبعد المنطقة عن متوسط المناطق الـ13.
- نعكس إشارة الدخل حتى تعني الدرجة الأعلى هشاشة أكبر في المؤشرات الثلاثة.
- **مؤشر الهشاشة = متوسط الدرجات الثلاث بأوزان متساوية.** نحسبه مرتين: بنسبة ضمان 2022، وبمتوسط 2020-2022.

**لماذا أوزان متساوية؟** لا يوجد مصدر يحدد أهمية كل مؤشر، ومع 13 منطقة فقط لا تصلح الطرق الإحصائية لاستخراج الأوزان. لذلك نعطي المؤشرات وزنًا متساويًا، ونختبر أثر تغييره في 8.1 (حذف مؤشر) و8.2 (ألف توزيع عشوائي للأوزان).

**احتمال تكرار الدخل وحصة الغذاء:** طُرح احتمال أن الدخل وحصة الغذاء يقيسان الشيء نفسه، لأن الأسر الأقل دخلًا تنفق حصة أكبر على الغذاء. ولو صح ذلك لأخذ هذا الجانب وزنًا مضاعفًا في المؤشر المركب. لكن ارتباطهما (0.64) ليس أعلى من ارتباط نسبة الضمان بحصة الغذاء (0.71)، فالعلاقة بينهما ليست استثنائية. كما أن حذف أي منهما في 8.1 لم يغيّر المناطق الخمس الأولى، فأبقينا المؤشرين.

```python
def zscore(s):
    return (s - s.mean()) / s.std(ddof=0)


def fragility_parts(df, ss_col):
    """درجات المؤشرات الثلاثة، والقيمة الأعلى تعني هشاشة أكبر."""
    return pd.DataFrame({
        "ss_rate": zscore(df[ss_col]),
        "low_income": -zscore(df["income_median_saudi"]),
        "food_share": zscore(df["food_share"]),
    })


parts_2022 = fragility_parts(ind, "ss_rate_2022")
ind["fragility_2022"] = parts_2022.mean(axis=1)
ind["fragility_avg"] = fragility_parts(ind, "ss_rate_avg").mean(axis=1)
ind.to_csv(OUTPUTS / "indicators.csv", index=False)  # بعد إضافة المؤشرين المركبين

print(
    parts_2022.assign(region=ind["region_en"], fragility_2022=ind["fragility_2022"], fragility_avg=ind["fragility_avg"])
    .set_index("region")
    .sort_values("fragility_2022", ascending=False)
    .round(2)
    .to_string()
)
print(f"\nارتباط ترتيب المؤشر بين 2022 والمتوسط: {ind['fragility_2022'].corr(ind['fragility_avg'], method='spearman'):.2f}")
```

الدرجة الموجبة تعني هشاشة أعلى من متوسط المناطق الـ13، والسالبة أقل. الأعمدة الثلاثة الأولى تبين أي مؤشر يرفع هشاشة كل منطقة أو يخفضها.

```python
# مؤشر الهشاشة: البرتقالي أهش من متوسط المناطق
d = ind.sort_values("fragility_2022").reset_index(drop=True)
fig, ax = plt.subplots(figsize=(7.5, 5))
ax.barh(range(len(d)), d["fragility_2022"], color=[ORANGE if v > 0 else OTHER for v in d["fragility_2022"]], height=0.65)
for yi, v in enumerate(d["fragility_2022"]):
    ax.text(v + (0.04 if v >= 0 else -0.04), yi, f"{v:+.2f}", va="center", ha="left" if v >= 0 else "right",
            fontsize=8.5, color=INK)
ax.axvline(0, color=MUTED, linewidth=1)
ax.set_yticks(range(len(d)), d["region_en"], color=INK)
ax.set_xlim(d["fragility_2022"].min() - 0.45, d["fragility_2022"].max() + 0.45)
ax.set_xlabel("Fragility index (0 = average of the 13 regions; higher = more fragile)", color=MUTED, fontsize=9)
clean_y(ax)
headline(fig, f"{d['region_en'].iloc[-1]} is the most fragile region, {d['region_en'].iloc[0]} the least",
         "Average of three z-scores: social security rate, Saudi median income (reversed), food share. "
         "Orange = more fragile than average", y=1.0, sub_y=0.94)
fig.savefig(FIGS / "fig_02_fragility_index.png")
plt.show()
```

## 7. مناطق الأولوية

**السؤال:** أي المناطق تجمع بين هشاشة أعلى وعدد منظمات أقل؟

**التعريف:** نرتّب المناطق بـ **درجة الفجوة** = الهشاشة المعيارية − الكثافة المعيارية. كلما ارتفعت كانت الهشاشة أكبر قياسًا بعدد المنظمات. المنطقة ذات الأولوية هي التي تبقى ضمن الخمس الأولى في الفجوة مهما غيّرنا طريقة الحساب (القسم 8).

**مصدر الفجوة:** الفجوة مجموع جزئين: جزء الهشاشة (الهشاشة المعيارية) وجزء نقص المنظمات (الكثافة المعيارية بإشارة معكوسة). نقارن الجزئين لنعرف هل دخلت المنطقة الأولوية بسبب الهشاشة أساسًا أم بسبب قلة المنظمات، دون عتبة حادة.

**الأرباع للوصف فقط:** نقسم المناطق بخطي الوسيط إلى أربع فئات (هشاشة عالية أو منخفضة × منظمات قليلة أو كثيرة) لوصف موقع كل منطقة، ولا نبني عليها القرار. السبب: مع 13 منطقة يقلب فرق صغير جدًا تصنيف المنطقة بالكامل، أما الفجوة فتقيس المسافة لا الجهة فقط.

**السيناريوهات الأربعة:**

| السيناريو | الضمان | الجهات |
|---|---|---|
| A | 2022 | الجمعيات والمؤسسات |
| B | 2022 | الجمعيات فقط |
| C | متوسط 2020-2022 | الجمعيات والمؤسسات |
| D | متوسط 2020-2022 | الجمعيات فقط |

```python
SCENARIOS = {
    "A": ("fragility_2022", "density_all"),
    "B": ("fragility_2022", "density_assoc"),
    "C": ("fragility_avg", "density_all"),
    "D": ("fragility_avg", "density_assoc"),
}


def gap_score(df, frag_col, dens_col):
    return zscore(df[frag_col]) - zscore(df[dens_col])


def quadrant(df, frag_col, dens_col):
    """موقع المنطقة بالنسبة لخطي الوسيط، للوصف فقط."""
    high_frag = df[frag_col] > df[frag_col].median()
    low_dens = df[dens_col] < df[dens_col].median()
    q = pd.Series("هشاشة منخفضة، منظمات كثيرة", index=df.index)
    q[high_frag & low_dens] = "هشاشة عالية، منظمات قليلة"
    q[high_frag & ~low_dens] = "هشاشة عالية، منظمات كثيرة"
    q[~high_frag & low_dens] = "هشاشة منخفضة، منظمات قليلة"
    return q


print(f"السيناريو A: وسيط الهشاشة {ind['fragility_2022'].median():.2f} | وسيط الكثافة {ind['density_all'].median():.2f}\n")

results = ind[["region_id", "region_en"]].copy()
for name, (frag_col, dens_col) in SCENARIOS.items():
    gap = gap_score(ind, frag_col, dens_col)
    results[f"{name}_gap"] = gap.round(2)
    results[f"{name}_gap_rank"] = gap.rank(ascending=False, method="min").astype(int)  # 1 = أكبر فجوة
# مصدر الفجوة في السيناريو A: أي الجزأين أكبر
frag_col, dens_col = SCENARIOS["A"]
results["A_frag_part"] = zscore(ind[frag_col]).round(2)
results["A_dens_part"] = (-zscore(ind[dens_col])).round(2)
results["A_source"] = np.where(results["A_frag_part"] > results["A_dens_part"], "الهشاشة", "نقص المنظمات")

rank_cols = [f"{s}_gap_rank" for s in SCENARIOS]
results["top5_scenarios"] = (results[rank_cols] <= 5).sum(axis=1)
results["A_quadrant"] = quadrant(ind, *SCENARIOS["A"])
results = results.sort_values("A_gap", ascending=False)

show_table(results[["region_en", "A_gap", "A_frag_part", "A_dens_part", "A_source", *rank_cols, "top5_scenarios"]],
           gradient=["A_gap"])
print("الأرباع (للوصف فقط):")
show_table(results[["region_en", "A_quadrant"]])

results.to_csv(OUTPUTS / "priority_results.csv", index=False, encoding="utf-8-sig")
```

## 8. اختبار الحساسية

**السؤال:** إلى أي مدى يتغير ترتيب المناطق عند تعديل طريقة الحساب؟

تهدف اختبارات الحساسية إلى قياس مدى استقرار ترتيب المناطق عند تغيير الاختيارات المنهجية. لكنها لا تختبر دقة البيانات الأصلية أو خلوّها من أخطاء القياس؛ إذ تُناقش هذه القيود في القسم 10.

غطّت السيناريوهات الأربعة في القسم 7 الاختلاف في سنة بيانات الضمان ونوع الجهات المحتسبة. وفي هذا القسم، نختبر بقية الاختيارات باستخدام السيناريو A، من خلال ثلاثة اختبارات:
- **8.1 التجارب المحددة:** تغيير اختيار منهجي واحد في كل مرة.
- **8.2 الأوزان العشوائية:** تغيير أوزان المؤشرات الثلاثة ألف مرة.
- **8.3 حذف منطقة واحدة في كل مرة:** إعادة الحساب بعد استبعاد كل منطقة بالتتابع، لأن المتوسط والانحراف المعياري يعتمدان على المناطق الداخلة في الحساب.

### 8.1 التجارب المحددة

تختبر هذه التجارب ثلاثة تغييرات:
- استخدام الرتب بدلًا من الدرجات المعيارية `z-score`.
- حذف مؤشر واحد في كل مرة.
- إضافة هامش الدخل بوصفه مؤشرًا رابعًا.

**معيار ثبات الأولوية:** تُعد المنطقة ذات أولوية مستقرة إذا ظهرت ضمن المراتب الخمس الأولى في خمس تجارب على الأقل من أصل ست.

```python
# كل تجربة: المؤشرات واتجاهها (+1 أعلى = أهش، -1 أقل = أهش)، والسيناريو A أساس المقارنة
BASE = {"ss_rate_2022": 1, "income_median_saudi": -1, "food_share": 1}
VARIANTS = {
    "base": BASE,
    "no_ss": {"income_median_saudi": -1, "food_share": 1},
    "no_income": {"ss_rate_2022": 1, "food_share": 1},
    "no_food": {"ss_rate_2022": 1, "income_median_saudi": -1},
    "plus_margin": {**BASE, "margin_saudi": -1},
}


def gap_z(df, components, dens_col="density_all"):
    frag = pd.concat([sign * zscore(df[col]) for col, sign in components.items()], axis=1).mean(axis=1)
    return zscore(frag) - zscore(df[dens_col])


def gap_ranks(df, components, dens_col="density_all"):
    """نفس الفكرة بالرتب: متوسط رتب المؤشرات بدل الدرجات المعيارية."""
    frag = pd.concat([(sign * df[col]).rank(pct=True) for col, sign in components.items()], axis=1).mean(axis=1)
    return frag.rank(pct=True) - df[dens_col].rank(pct=True)


# كل الرتب بطريقة min: عند التعادل تأخذ المنطقتان الرتبة نفسها (مثل 6 و6)
sens = pd.DataFrame({"region": ind["region_en"]})
for name, comps in VARIANTS.items():
    sens[name] = gap_z(ind, comps).rank(ascending=False, method="min").astype(int)
sens["ranks_not_z"] = gap_ranks(ind, BASE).rank(ascending=False, method="min").astype(int)

variant_cols = [c for c in sens.columns if c != "region"]
n_tests = len(variant_cols)
sens["best"] = sens[variant_cols].min(axis=1)
sens["worst"] = sens[variant_cols].max(axis=1)
sens["top5_in_n"] = (sens[variant_cols] <= 5).sum(axis=1)
sens = sens.sort_values("base")


# الأولوية الثابتة: ضمن الخمس الأولى في كل التجارب إلا واحدة على الأكثر
priority = set(sens.loc[sens["top5_in_n"] >= n_tests - 1, "region"])
priority_order = list(sens.loc[sens["region"].isin(priority), "region"])  # بترتيب الفجوة في الأساس
show_table(sens, highlight=priority, region_col="region", reverse=variant_cols, precision=0)

print("ارتباط كل تجربة بالأساس (Spearman):")
print(sens[variant_cols].corr(method="spearman")["base"].drop("base").round(2).to_string())
print(f"\nالأولوية الثابتة (ضمن الخمس الأولى في {n_tests - 1} تجارب أو أكثر من {n_tests}):", ", ".join(priority_order))

sens.to_csv(OUTPUTS / "sensitivity_ranks.csv", index=False)
```

**قراءة الجدول:** يوضح كل عمود ترتيب المناطق وفق درجة الفجوة في تجربة محددة؛ حيث تمثل المرتبة 1 أكبر فجوة. ويعرض `best` و`worst` أفضل وأسوأ مرتبة حققتها المنطقة عبر التجارب، بينما يوضح `top5_in_n` عدد التجارب التي ظهرت فيها ضمن المراتب الخمس الأولى.

**النتائج:**
- جاءت نجران في المرتبة الأولى في جميع التجارب.
- حافظت المدينة المنورة وتبوك على وجودهما ضمن المراتب الخمس الأولى في جميع التجارب.
- ظهرت جازان ومكة المكرمة ضمن الخمس الأولى في خمس تجارب من أصل ست. تراجعت جازان إلى المرتبة السادسة عند حذف مؤشر الضمان، بينما تراجعت مكة إلى المرتبة السادسة عند استخدام الرتب (متعادلة مع المنطقة الشرقية).
- ظهرت الجوف ضمن الخمس الأولى في تجربتين فقط.
- تراوح ارتباط ترتيب المناطق في كل تجربة بالترتيب الأساسي بين 0.86 و0.97، ما يشير إلى أن النتائج لا تعتمد على اختيار منهجي واحد.

### 8.2 اختبار الأوزان العشوائية

يعطي المؤشر المركب المؤشرات الثلاثة أوزانًا متساوية. ولأن هذا الاختيار افتراض منهجي، نعيد حساب المؤشر ألف مرة، مع توزيع الأوزان عشوائيًا في كل مرة، بحيث يكون مجموعها واحدًا.

بعد ذلك، نحسب نسبة المرات التي ظهرت فيها كل منطقة ضمن المراتب الخمس الأولى. وتُثبّت البذرة العشوائية على القيمة `42` لضمان إمكانية إعادة إنتاج النتائج نفسها.

```python
rng = np.random.default_rng(42)
weights = rng.dirichlet(np.ones(len(BASE)), size=1000)  # كل صف: أوزان المؤشرات الثلاثة، ومجموعها 1
parts = pd.concat([sign * zscore(ind[col]) for col, sign in BASE.items()], axis=1)

draw_ranks = pd.concat(
    [(zscore(parts @ w) - zscore(ind["density_all"])).rank(ascending=False, method="min") for w in weights], axis=1
)
weights_test = pd.DataFrame({
    "region": ind["region_en"],
    "weights_top5_pct": (draw_ranks <= 5).mean(axis=1) * 100,
    "weights_median_rank": draw_ranks.median(axis=1),
    "weights_best": draw_ranks.min(axis=1).astype(int),
    "weights_worst": draw_ranks.max(axis=1).astype(int),
}).sort_values("weights_top5_pct", ascending=False)
show_table(weights_test, highlight=priority, region_col="region", gradient=["weights_top5_pct"], precision=1)
```

**قراءة الجدول:**
- `weights_top5_pct`: نسبة توزيعات الأوزان التي ظهرت فيها المنطقة ضمن الخمس الأولى.
- `weights_best`: أفضل مرتبة حققتها المنطقة.
- `weights_worst`: أسوأ مرتبة حققتها المنطقة.

**النتائج:**
- حافظت نجران والمدينة المنورة على وجودهما ضمن الخمس الأولى في 100% من توزيعات الأوزان.
- ظهرت مكة المكرمة ضمن الخمس الأولى في 96.2% من التوزيعات، وتبوك في 83.8%.
- ظهرت جازان ضمن الخمس الأولى في 72.9% فقط من التوزيعات، وتراوحت مرتبتها بين الأولى والثامنة. ويُعزى ذلك إلى اعتماد جزء كبير من هشاشتها على نسبة مستفيدي الضمان، ما يؤدي إلى تراجعها عندما ينخفض وزن هذا المؤشر.
- جاءت الجوف بعد ذلك بنسبة 27.5%، ثم عسير بنسبة 19.6%. ولم تظهر بقية المناطق ضمن الخمس الأولى في أي توزيع.

### 8.3 حذف منطقة واحدة في كل مرة

تعتمد الدرجات المعيارية على متوسط المناطق وانحرافها المعياري؛ لذلك قد تؤثر المناطق ذات القيم المتطرفة، مثل الرياض في الدخل أو الباحة في كثافة المنظمات، في ترتيب بقية المناطق.

لاختبار هذا الأثر، نحذف كل منطقة بالتتابع، ثم نعيد حساب درجة الفجوة وفق السيناريو A للمناطق الاثنتي عشرة المتبقية. ونحسب عدد المرات التي بقيت فيها كل منطقة ضمن الخمس الأولى.

**ملاحظة منهجية:** يؤدي حذف إحدى المناطق المتقدمة إلى إتاحة مكان جديد لمنطقة أخرى ضمن الخمس الأولى؛ لذلك نعرض أسوأ مرتبة حققتها كل منطقة أيضًا.

```python
loo_ranks = pd.DataFrame(index=ind.index, columns=ind.index, dtype=float)  # صف لكل منطقة، عمود لكل منطقة محذوفة
for dropped in ind.index:
    rest = ind.drop(index=dropped)
    loo_ranks.loc[rest.index, dropped] = gap_z(rest, BASE).rank(ascending=False, method="min")

leave_one_out = pd.DataFrame({
    "region": ind["region_en"],
    "loo_top5": (loo_ranks <= 5).sum(axis=1),  # من 12 مرة
    "loo_best": loo_ranks.min(axis=1).astype(int),
    "loo_worst": loo_ranks.max(axis=1).astype(int),
}).sort_values(["loo_top5", "loo_worst"], ascending=[False, True])
show_table(leave_one_out, highlight=priority, region_col="region", gradient=["loo_top5"], precision=0)

robust = weights_test.merge(leave_one_out, on="region")
robust.to_csv(OUTPUTS / "robustness_checks.csv", index=False)
```

**قراءة الجدول:**
- `loo_top5`: عدد المرات التي بقيت فيها المنطقة ضمن الخمس الأولى من أصل 12 تجربة.
- `loo_best`: أفضل مرتبة حققتها المنطقة.
- `loo_worst`: أسوأ مرتبة حققتها المنطقة.

**النتائج:**
- حافظت نجران والمدينة المنورة وجازان وتبوك على وجودها ضمن الخمس الأولى في جميع التجارب الاثنتي عشرة.
- ظهرت مكة المكرمة ضمن الخمس الأولى في 11 تجربة، وكانت أسوأ مرتبة لها السادسة.
- ظهرت الجوف ضمن الخمس الأولى في ست تجارب فقط.

## 10. القيود

### أولًا: قيود البيانات

**1. اختلاف الفترات الزمنية للبيانات:** لا تعود جميع المؤشرات إلى السنة نفسها؛ فبيانات السكان والضمان الاجتماعي تعود إلى 2022، وتشير بيانات الجمعيات إلى منتصف 2022 تقريبًا، بينما يعود مسح الدخل والإنفاق إلى 2023، وهو أحدث إصدار مستخدم في التحليل.

لا تتوافر بيانات منشورة موحّدة للسكان والضمان حسب المنطقة لعام 2023 تتيح توحيد الفترة الزمنية لجميع المؤشرات. ومع ذلك، يُرجّح أن يكون أثر هذا الاختلاف محدودًا للأسباب التالية:
- يركز التحليل على ترتيب المناطق نسبيًا، وقد أظهر ترتيبها في نسبة الضمان استقرارًا كبيرًا: بلغ الارتباط 0.99 بين 2020 و2022، و0.94 بين 2021 و2022 (القسم 4.3).
- توجد فروق كبيرة بين المناطق في بعض المؤشرات؛ فعلى سبيل المثال، تبلغ حصة الغذاء 43.5% في نجران مقابل 23.5% في الرياض.

**2. تقديرات أعداد السكان السعوديين لعامي 2020 و2021:** أعادت الهيئة العامة للإحصاء تقدير أعداد السكان السعوديين للسنوات السابقة بالاستناد إلى تعداد 2022. ولذلك، لا تمثل أرقام 2020 و2021 تعدادًا فعليًا مستقلًا، وتؤثر هذه المحدودية في السيناريوهات التي تستخدم متوسط الفترة 2020-2022 (C وD).

**3. حصة الغذاء تشمل جميع الأسر:** لا يوفر المصدر المستخدم تفصيلًا لحصة الغذاء حسب الجنسية داخل كل منطقة؛ لذلك استُخدمت حصة الغذاء لجميع الأسر بدلًا من الأسر السعودية وحدها. وتشير بيانات المسح على مستوى المملكة (الجدول 12-4 في ملف المسح) إلى تقارب هذه الحصة بين السعوديين وغير السعوديين، ما يرجّح محدودية أثر هذا الاختيار.

**4. عدم معرفة هامش الخطأ ودقة القياس:** يعتمد مسح الدخل والإنفاق على عينة، ولا يتوافر في البيانات المستخدمة تقدير واضح لهامش الخطأ. كما أن أعداد مستفيدي الضمان تعود إلى سنة انتقالية بين نظامين. ولذلك، لا يمكن استبعاد أن تكون بعض الفروق المحدودة بين المناطق، مثل مناطق المجموعة الثانية، ناتجة عن أخطاء القياس. وتختبر تحليلات الحساسية استقرار النتائج عند تغيير طريقة الحساب، لكنها لا تثبت صحة الأرقام الأصلية.

### ثانيًا: قيود المؤشرات

**1. نسبة الضمان في سنة انتقالية:** شهد عام 2022 الانتقال إلى نظام الضمان الاجتماعي المطور، الذي يعتمد شروط أهلية جديدة. لذلك، تعبّر نسبة المستفيدين في هذه السنة جزئيًا عن تغطية النظام وإجراءات الانتقال إليه، وليس عن مستوى الفقر وحده. أما متوسط 2020-2022، فيجمع بيانات من فترتين تختلفان في نظام الضمان المطبق.

**2. كثافة المنظمات ليست مقياسًا مباشرًا للخدمات الفعلية:** يعتمد المؤشر على عدد المنظمات المسجلة مقارنة بعدد السكان، لكنه لا يقيس مباشرةً حجم الخدمات المقدمة أو مدى وصولها إلى المستفيدين. وتظهر هذه المحدودية في ثلاثة جوانب:
- **المقر لا يحدد نطاق الخدمة:** قد تقدم الجمعية المسجلة في الرياض خدماتها لمستفيدين في مناطق أخرى.
- **العدد لا يعكس حجم المنظمة:** تُحسب الجمعية الكبيرة والصغيرة بوصفهما منظمتين، دون مراعاة اختلاف مواردهما أو قدرتهما التشغيلية.
- **المؤشر لا يشمل جميع مصادر الخدمة:** لا تدخل جميع أشكال الدعم الحكومي والعمل الخيري غير المسجل في الحساب.

**3. أوزان متساوية لمكوّنات الفجوة:** تعطي المنهجية الهشاشة الاقتصادية ونقص كثافة المنظمات وزنًا متساويًا في حساب درجة الفجوة. وقد تدخل منطقة ضمن الأولوية بسبب انخفاض كثافة المنظمات وحده، كما هو الحال في مكة المكرمة. لذلك، لا ينبغي تفسير ترتيب الفجوة بمعزل عن مصدرها؛ بل يجب التمييز بين المناطق التي تعاني من هشاشة اقتصادية مرتفعة، والمناطق التي تتمثل حاجتها الأساسية في سد نقص الخدمات المؤسسية (القسم 9).

**الخلاصة النهائية:** تظل التوصيات استدلالات مبنية على مؤشرات إقليمية وبيانات ذات قيود، وينبغي دعمها ببيانات مباشرة عن حجم الخدمات المقدمة، وموارد المنظمات، واحتياجات المستفيدين الفعلية قبل اتخاذ قرارات نهائية بشأن توزيع التمويل.
